# Split into train & test :

```python
X = df.drop(["claim", "Id"], axis=1)
y = df["claim"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# imputation :

```python
X_train.isna().sum()
```

```python
X_train["age"] = X_train["age"].fillna(X_train["age"].mean())
X_test["age"] = X_test["age"].fillna(X_train["age"].mean())


X_train["region"] = X_train["region"].fillna(X_train["region"].mode()[0])
X_test["region"] = X_test["region"].fillna(X_train["region"].mode()[0])
```

```python
X_train.isna().sum()
```

```python
X_train.head(10)
```

# Scaling :

```python
num_train = X_train.select_dtypes("number")
num_test = X_test.select_dtypes("number")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_num_train = scaler.fit_transform(num_train)
scaled_num_test = scaler.transform(num_test)
```

```python
num_train = pd.DataFrame(scaled_num_train, columns=num_train.columns)
num_test = pd.DataFrame(scaled_num_test, columns=num_test.columns)
```

```python
num_train
```

# Encoding :

```python
cat_train = X_train.select_dtypes("object")
cat_test = X_test.select_dtypes("object")
```

```python
cat_train = pd.get_dummies(cat_train, drop_first=True)
cat_test = pd.get_dummies(cat_test, drop_first=True)
cat_test = cat_test.reindex(columns=cat_train.columns, fill_value=0)
```

```python
cat_train
```

# concat numerical and categorical data :

```python
num_train = num_train.reset_index(drop=True)   # **i have to reset their indexes, because if i don't it will concat by index, so it causes missing values after concatenation !!!**
cat_train = cat_train.reset_index(drop=True)

num_test = num_test.reset_index(drop=True)
cat_test = cat_test.reset_index(drop=True)
```

```python
new_X_train = pd.concat([num_train, cat_train], axis=1)
new_X_test = pd.concat([num_test, cat_test], axis=1)

X_train = new_X_train.copy()
X_test = new_X_test.copy()
```

```python
new_X_train
```

```python
X_train = new_X_train.copy()
X_test = new_X_test.copy()
```

# training & Evaluation :

## Linear Regression :

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train, y_train)

y_pred = lr_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"mse : {mse} \n r2 : {r2}")
```

## Polynomial Features :

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

pf_model = LinearRegression()
poly = PolynomialFeatures(degree=2)
poly_X_train = poly.fit_transform(X_train)

pf_model.fit(poly_X_train, y_train)

y_pred = pf_model.predict(poly.transform(X_test))
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"mse : {mse} \n r2 : {r2}")
```

## Random Forest :

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor()

rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"mse : {mse} \n r2 : {r2}")
```

# Hyperparameters tuning :

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    random_state=42
)

params = {
    "n_estimators": [100, 200, 300],
    "max_depth": [3, 5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10, 20],
    "min_samples_leaf": [1, 2, 5, 10],
    "max_features": [0.5, 0.7, 1.0]
}

search = RandomizedSearchCV(
    rf,
    params,
    n_iter=20,
    cv=5,
    scoring="r2",
    random_state=42,
    n_jobs=-1
)

search.fit(X_train, y_train)

print(search.best_params_)
print(search.best_score_)
```

```python
best_rf = search.best_estimator_

y_pred = best_rf.predict(X_test)

print(r2_score(y_test, y_pred))
```
